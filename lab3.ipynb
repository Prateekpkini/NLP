```python
def min_edit_distance_matrix(str1,str2):
    m,n=len(str1),len(str2)
    dp=[[0]*(n+1) for i in range (m+1)]
    for i in range (m+1):
        dp[i][0]=i
    for j in range (n+1):
        dp[0][j]=j
    for i in range (1,m+1):
        for j in range(1,n+1):
            if str1[i-1]==str2[j-1]:
                dp[i][j]=dp[i-1][j-1]
            else:
                dp[i][j]=1+min(dp[i-1][j],dp[i][j-1],dp[i-1][j-1])
    return dp
str1_example="saturday"
str2_example="sunday"
dp_matrix=min_edit_distance_matrix(str1_example,str2_example)

for row in dp_matrix:
    print(row)
edit_distance=dp_matrix[len(str1_example)][len(str2_example)]
print(f"The minimum edit distance between '{str1_example}' and '{str2_example}' is {edit_distance}")
```
