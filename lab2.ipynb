```python
import nltk
from nltk import word_tokenize
from collections import Counter,defaultdict
import math
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
paragraph="The cat sat on the mat. The cat ate the fish. The dog barked at the cat. A cat sleeps all day. A dog runs in the park." 
```

```python
sentences=nltk.sent_tokenize(paragraph.lower())
tokenized_sentences=[word_tokenize(s.lower()) for s in sentences]
all_tokens=[token for sent in tokenized_sentences for token in sent]
```

```python
unigram_counts=Counter(all_tokens)
def unigram_prob(word):
    return unigram_counts[word]/len(all_tokens)
```

```python
bigram_counts=Counter()
for sent in tokenized_sentences:
    for i in range (len(sent)-1):
        bigram_counts[(sent[i],sent[i+1])]+=1
def bigram_prob(w1,w2):
    return bigram_counts[(w1,w2)]/unigram_counts[w1]
```

```python
trigram_counts=Counter()
for sent in tokenized_sentences:
    for i in range (len(sent)-2):
        trigram_counts[(sent[i],sent[i+1],sent[i+2])]+=1
def trigram_prob(w1,w2,w3):
    return trigram_counts[(w1,w2,w3)]/bigram_counts[(w1,w2)]
```

```python
print("Unigram probability of 'mat':",unigram_prob("mat"))
print("Bigram probability of ('cat','ate'):",bigram_prob("cat","ate"))
print("Trigram probability of ('cat','sleeps','all'):",trigram_prob("cat","sleeps","all"))
```

```python
test_sentence="The cat sat on the mat"
test_tokens=word_tokenize(test_sentence.lower())
uni_prob=1
for word in test_tokens:
    uni_prob=unigram_prob(word)
bi_prob=unigram_prob(test_tokens[0])
for i in range (len(test_tokens)-1):
    bi_prob*=bigram_prob(test_tokens[i],test_tokens[i+1])
```

```python
tri_prob=unigram_prob(test_tokens[0])*bigram_prob(test_tokens[0],test_tokens[1])
for i in range(len(test_tokens)-2):
    tri_prob*=trigram_prob(test_tokens[i],test_tokens[i+1],test_tokens[i+2])
```

```python
print("Sentence:", test_sentence)
print("Unigram Probability:",uni_prob)
print("Bigram PRobability:",bi_prob)
print("Trigram Probability:",tri_prob)
```
