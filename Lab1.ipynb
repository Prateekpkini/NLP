```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import unicodedata
```

```python
nltk.download('punkt')
```

```python
nltk.download('stopwords')
```

```python
def preprocess(text):
    tokens= word_tokenize(text)
    print("\n Tokenized words: \n",tokens)
    filtered_tokens=[word for word in tokens if word.isalpha()]
    print("\n Filtered tokens (no special characters/numbers):\n",filtered_tokens)
    validated_tokens=[word for word in filtered_tokens if is_english(word)]
    print("\n Script validated tokens (English only): \n",validated_tokens)
    stop_words=set(stopwords.words('english'))
    stopword_removed_tokens=[word for word in validated_tokens if word.lower() not in stop_words]
    print("\n Stop words removed:\n",stopword_removed_tokens)
    stemmer=PorterStemmer()
    stemmed_tokens=[stemmer.stem(word.lower()) for word in stopword_removed_tokens]
    print("\n Stemmed tokens:\n", stemmed_tokens)
    return stemmed_tokens
```

```python
def is_english(word):
    return all('A'<= char<='Z' or 'a'<= char <='z' for char in word)
```

```python
text= ("The NLP system is analysing various preprocessing tasks such as \"tokenisation, filteration, script validation, stop word removal \" Stemming on 1-1-2026")
```

```python
preprocessed_text=preprocess(text)
print("\n Final preprocessed output : \n",preprocessed_text)
```
