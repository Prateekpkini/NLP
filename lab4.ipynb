```python
tokens=[]
pos=0
```

```python
def match(expected):
    global pos
    if pos<len(tokens) and tokens[pos].lower() == expected.lower():
        pos+=1
        return True
    return False
```

```python
def SingularNoun():
    return match("flight")
def NOM():
    return SingularNoun()
def Det():
    return match("that")
def NP():
    return Det() and  NOM()
def Verb():
    return match("book")
def VP():
    return Verb() and NP()
def S():
    return VP()
```

```python
def parse_topdown(input_tokens):
    global tokens,pos
    tokens=input_tokens
    pos=0
    return S() and pos==len(tokens)
    return S() and pos==len(tokens)
```

```python
sentence=["Book", "that", "flight"]
print("Top-Down Parsing:", parse_topdown(sentence))
```

```python
grammar={
    ("book",):"Verb",
    ("that",):"Det",
    ("flight",):"SingularNoun",
    ("SingularNoun",):"NOM",
    ("Det","NOM"):"NP",
    ("Verb","NP"):"VP",
    ("VP",):"S"
}
```

```python
def reduce(stack):
    for length in range(3,0,-1):
        if tuple (stack[-length:]) in grammar :
            non_terminal=grammar[tuple(stack[-length:])]
            del stack[-length:]
            stack.append(non_terminal)
            return True
    return False
```

```python
def parse_bottomup (input_tokens):
    stack=[]
    for token in input_tokens:
        stack.append(token.lower())
        while reduce (stack):
            pass
    return stack==["S"]
```

```python
sentence=["Book","that","flight"]
print("Bootom-Up parsing:",parse_bottomup(sentence))
```
