```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch
```

```python
# Load a pretrained multilingual model (Example: English to kannada)
model_name = "rajaykumar12959/nllb-en-kn-v1"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
# Example sentence
src_texts = ["Artificial intelligence is transforming education."]
```

```python
# Tokenize
inputs = tokenizer(src_texts, return_tensors="pt", padding=True, truncation=True)
```

```python
# Translate
translated = model.generate(**inputs)
tgt_texts = [tokenizer.decode(t, skip_special_tokens=True) for t in translated]
```

```python
print("Translation:", tgt_texts)
```
