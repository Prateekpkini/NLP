```python
import nltk
from nltk.corpus import brown,inaugural,reuters,udhr
```

```python
nltk.download('brown')
nltk.download('inaugural')
nltk.download('reuters')
nltk.download('udhr')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
print("="*70)
print("1.BROWN CORPUS")
print("="*70)
      
```

```python
print("Brown fileds (first 10):")
print(brown.fileids()[:10])
```

```python
print("\n Brown categories:")
print(brown.categories()[:20])
```

```python
print("\n Brown raw text from file (first 100 characters):")
brown_file=brown.fileids()[0]
print(brown.raw(brown_file)[:300])
```

```python
print("\n Brown words from first file(first 20 words):")
print(brown.words(brown_file)[:20])
```

```python
print("\n Brown sentences from first file(first 2 sentences):")
print(brown.sents(brown_file)[:2])
```

```python
search_word="government"
brown_words=brown.words(categories='news')
count_brown=brown_words.count(search_word)
print(f"Frequency of (search_word)' in Brown news category:", count_brown)
```
